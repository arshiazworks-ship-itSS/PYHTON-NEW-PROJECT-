```python
candies = {
    "A1": {"candy_name": "chocolate bar", "candy_price": 100, "candy_quantity": 5},
    "A2": {"candy_name": "gummy bear", "candy_price": 80, "candy_quantity": 0},
    "A3": {"candy_name": "lollipop", "candy_price": 30, "candy_quantity": 10},
}
candies["A4"] = {"candy_name": "candy_puffs", "candy_price": 50, "candy_quantity": 6}
candies["A3"]["candy_price"] = 49   # update only the price
```

# FUNCTIONS 

```python
def calculate(candy_price, candy_quantity):
    total = candy_price * candy_quantity
    if candy_quantity >= 5:
        total = total - (total * 0.10)   # 10% discount
    return total
```

```python
def avail_stock(stock):
    if stock > 0:
        return "in stock"
    else:
        return "out of stock"
```

# INPUT 

```python
candy_name = input("Enter the name: ")
candy_code = input("Enter the code: ")

# keep asking until the price is a valid positive number
```

```python
while True:
    try:
        candy_price = int(input("Enter the price: "))
        if candy_price > 0:
            print("Price is valid")
            break
        else:
            print("The price is invalid, try again")
    except ValueError:
        print("Please enter a number only")

# keep asking until the quantity is a valid whole number
while True:
    try:
        candy_quantity = int(input("Enter the quantity: "))
        if candy_quantity >= 0:
            break
        print("Quantity can't be negative")
    except ValueError:
        print("Non-integer values are invalid")

print(candy_name.title())
print("Total:", calculate(candy_price, candy_quantity))
print(avail_stock(candy_quantity))
print(candies)
```
